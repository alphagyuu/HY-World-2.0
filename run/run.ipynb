## 🚀 Step 2: 3D 월드 생성 (Run Pipeline)
이미지 경로와 월드 이름을 설정하고 실행하세요.

```python
import os
from run.main import WorldGenerator

# --- 설정 변수 (사용자 입력) ---
INPUT_IMAGE = "input.png"   # @param {type:"string"}
WORLD_NAME = "MyNewWorld"   # @param {type:"string"}
MODE = "perspective"        # @param ["perspective", "panorama"]

if not os.path.exists(INPUT_IMAGE):
    print(f"[!] '{INPUT_IMAGE}' 파일을 찾을 수 없습니다. 경로를 확인해주세요.")
else:
    generator = WorldGenerator()
    print(f"[*] Starting {MODE} generation...")
    result = generator.generate(INPUT_IMAGE, WORLD_NAME, mode=MODE)

    if result["status"] == "success":
        print(f"\n[🎉] 생성 완료! 결과 폴더: {result['output_path']}")
    else:
        print(f"\n[❌] 실패: {result['message']}")
```
